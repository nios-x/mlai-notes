```python
import numpy as np

# Exam A
examA = np.array([40, 50, 60])

# Exam B
examB = np.array([400, 500, 600])

# Normalize (Z-score normalization)
normA = (examA - np.mean(examA)) / np.std(examA)
normB = (examB - np.mean(examB)) / np.std(examB)

print("Exam A")
print("Marks      :", examA)
print("Mean       :", np.mean(examA))
print("Std        :", np.std(examA))
print("Normalized :", normA)

print()

print("Exam B")
print("Marks      :", examB)
print("Mean       :", np.mean(examB))
print("Std        :", np.std(examB))
print("Normalized :", normB)

print()

print("Are the normalized arrays equal?")
print(np.allclose(normA, normB))
```

```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 100

df = pd.DataFrame({
    "id": np.arange(1, n + 1),
    "name": np.random.choice(
        ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Helen"],
        n
    ),
    "age": np.random.randint(18, 65, n),
    "salary": np.random.normal(60000, 15000, n).round(2),
    "department": np.random.choice(
        ["HR", "IT", "Sales", "Finance", "Marketing", "Support"],
        n
    ),
    "experience": np.random.randint(0, 41, n),
    "rating": np.random.uniform(1, 5, n).round(2),
    "is_manager": np.random.choice([True, False], n),
})

print(df.head())
```

```python
import pandas as pd

employees = pd.DataFrame({
    "emp_id": [101, 102, 103, 104, 105],
    "name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "department_id": [1, 2, 1, 3, 4]
})

departments = pd.DataFrame({
    "department_id": [1, 2, 3, 5],
    "department_name": ["HR", "IT", "Finance", "Marketing"],
    "location": ["Delhi", "Mumbai", "Bangalore", "Chennai"]
})

print(employees)
print()
print(departments)
```

```python
pd.merge(employees, departments, on="department_id", how="outer" )
pd.merge(employees, departments, on="department_id", how="inner" )
pd.merge(employees, departments, on="department_id", how="left" )
pd.merge(employees, departments, on="department_id", how="right" )
```
