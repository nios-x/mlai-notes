```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Load MNIST dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

X_train = X_train[..., tf.newaxis]
X_test = X_test[..., tf.newaxis]

print(X_train.shape)
print(X_test.shape)

model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.5),
    layers.Dense(10, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.1
)

test_loss, test_accuracy = model.evaluate(X_test, y_test)

print(f"\nTest Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")
```

```python
model.save("mnist_cnn.keras")
```

```python
import tkinter as tk
from PIL import Image, ImageDraw, ImageOps
import numpy as np
import tensorflow as tf

# Load trained model
model = tf.keras.models.load_model("mnist_cnn.keras")

WIDTH = 280
HEIGHT = 280
BG = "black"
FG = "white"

root = tk.Tk()
root.title("MNIST Digit Predictor")

canvas = tk.Canvas(root, width=WIDTH, height=HEIGHT, bg=BG)
canvas.pack()

image = Image.new("L", (WIDTH, HEIGHT), color=0)
draw = ImageDraw.Draw(image)

last_x = None
last_y = None


def start_draw(event):
    global last_x, last_y
    last_x, last_y = event.x, event.y


def draw_digit(event):
    global last_x, last_y

    if last_x is not None:
        canvas.create_line(
            last_x,
            last_y,
            event.x,
            event.y,
            fill=FG,
            width=18,
            capstyle=tk.ROUND,
            smooth=True,
        )

        draw.line(
            [last_x, last_y, event.x, event.y],
            fill=255,
            width=18,
        )

    last_x, last_y = event.x, event.y


def reset(event):
    global last_x, last_y
    last_x = None
    last_y = None


canvas.bind("<Button-1>", start_draw)
canvas.bind("<B1-Motion>", draw_digit)
canvas.bind("<ButtonRelease-1>", reset)


prediction_label = tk.Label(root, text="Prediction: ", font=("Arial", 18))
prediction_label.pack()


def predict():
    img = image.resize((28, 28))
    img = ImageOps.invert(img)

    arr = np.array(img).astype("float32") / 255.0
    arr = arr.reshape(1, 28, 28, 1)

    pred = model.predict(arr, verbose=0)

    digit = np.argmax(pred)
    confidence = np.max(pred) * 100

    prediction_label.config(
        text=f"Prediction: {digit} ({confidence:.2f}%)"
    )


def clear():
    canvas.delete("all")
    draw.rectangle([0, 0, WIDTH, HEIGHT], fill=0)
    prediction_label.config(text="Prediction: ")


tk.Button(root, text="Predict", command=predict).pack(pady=5)
tk.Button(root, text="Clear", command=clear).pack()

root.mainloop()
```
