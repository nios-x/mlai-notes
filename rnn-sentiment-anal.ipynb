```python
import tensorflow as tf
imdb=tf.keras.datasets.imdb
```

```python
(X_train, y_train), (X_test, y_test) = imdb.load_data()
```

```python
word_index = imdb.get_word_index()
reverse_word_index = {v: k for k, v in word_index.items()}
```

```python
#ID	Meaning
#0	<PAD>
#1	<START>
#2	<UNK> (unknown word)
#3	<UNUSED>
# " ".join([reverse_word_index.get(x-3,"?")  for x in X_train[0]])
```

```python
from tensorflow.keras.preprocessing.text import one_hot
vocab_size = 10000
sentences = []
for i in range(len(X_train)):
    sentences.append(" ".join([reverse_word_index.get(x-3,"?")  for x in X_train[i]]))
```

```python
tokens=[]
for x in sentences:
    tokens.append(one_hot(x, vocab_size))
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
sent_len = 500  
padded_reviews = pad_sequences(
    tokens,
    maxlen=sent_len,
    padding="pre",          # add zeros at the beginning
    truncating="pre"     # remove words from the beginning if too long
)
```

```python
  padded_reviews[0]
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=128,),
    SimpleRNN(128),
    Dense(1, activation="sigmoid")
])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
history = model.fit(
    padded_reviews,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.2
)
```

```python
from tensorflow.keras.preprocessing.text import one_hot
from tensorflow.keras.preprocessing.sequence import pad_sequences

def predict(x):
    encoded = one_hot(x, vocab_size)
    padded = pad_sequences(
        [encoded],
        maxlen=500,
        padding="pre",
        truncating="pre"
    )
    prediction = model.predict(padded)[0][0]
    print("Probability:", prediction)
    if prediction >= 0.5:
        print("Positive Review 😊")
    else:
        print("Negative Review 😞")
```
